# Hotel Booking EDA

## Data Preprocessing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('hotel_bookings.csv')
df.head()
```

```python
df.columns
```

```python
df[['children', 'babies', 'adults']].head(15)
```

This tells us the count of adults, babies and children. If there are rows will all 3 values as zeroes, then we need to remove them as it does not make any sense. 

I have removed my unwanted values from the df dataframe.
To achieve this use boolean indexing directly. we can negate 'unwanted' variable with ~ and filter df like this:

```python
unwanted = (df['children'] == 0) & (df['adults'] == 0) & (df['babies'] == 0)
unwanted
```

```python
df = df[~unwanted]
df
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace = True)
```

```python
df.shape
```

```python
uncancelled_booking = df[df['is_canceled']==0]
uncancelled_booking.head()
```

```python
uc_country = uncancelled_booking['country'].value_counts().reset_index()
uc_country.head(10)
```

```python
uc_country.columns = ['Country', 'No of guests']
uc_country
```

## Understanding the nature of room types - Reserved vs assigned

```python
pd.crosstab(index = uncancelled_booking['reserved_room_type'],columns = uncancelled_booking['assigned_room_type'],margins = True)
```
